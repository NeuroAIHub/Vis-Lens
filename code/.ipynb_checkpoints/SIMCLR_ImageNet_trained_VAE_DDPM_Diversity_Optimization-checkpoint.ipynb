```python
def test_input_for_unit_activation(net, layer_name, unit_idx, input_images):
    net.eval()
    target_activation = [None]

    def hook(module, input, output):
        if isinstance(unit_idx, tuple):
            
            activation = torch.mean(output, dim=(2, 3)) 
            target_activation[0] = activation
        else:
            target_activation[0] = output[:, unit_idx].mean(1).mean(1)

    # Register the hook to capture activations
    layers = {
        'layer1_0_conv1': net.encoder.model.convnet.layer1[0].conv1,
        'layer1_0_conv2': net.encoder.model.convnet.layer1[0].conv2,
        'layer2_2_conv2': net.encoder.model.convnet.layer2[2].conv2,
    }
    layer = layers[layer_name]
    hook_handle = layer.register_forward_hook(hook)
    with torch.no_grad():
        _ = net.encoder(input_images) 
    hook_handle.remove()  
    return target_activation[0]


device = torch.device('cuda:2' if torch.cuda.is_available() else 'cpu')
data_dir = '/data/local/yisong/test/imagenet-mini/val'
transform = transforms.Compose(
                [transforms.RandomResizedCrop(128, scale=(0.08, 1.0)),
                transforms.RandomHorizontalFlip(),
                transforms.ToTensor()]
                )
dataset = datasets.ImageFolder(root=data_dir, transform=transform)
data_loader = DataLoader(dataset, batch_size=100, shuffle=False, num_workers=4)
with torch.no_grad():
	test_feature_maps = []
	images = []
	for batch in data_loader:
		input_image = batch[0].to(device)
		activation_loss = test_input_for_unit_activation(
            ANN_model.to(device),
            layer_name='layer1_0_conv1',
            unit_idx=(5, 5),
            input_images=input_image
        )
        # Gather activations across all GPUs
		test_feature_maps.append(activation_loss.cpu())
		images.append(input_image.cpu())
	
	# Concatenate feature maps after gathering from all GPUs
	test_feature_maps_tensor = torch.cat(test_feature_maps, dim=0)
	images = torch.cat(images, dim=0)

print("All images:", images.shape)
print("Activation loss shape:", test_feature_maps_tensor.shape)
print("Max loss:", torch.max(test_feature_maps_tensor))
print("Min loss:", torch.min(test_feature_maps_tensor))
```

```python
class CustomDataset(Dataset):
    def __init__(self, images, activations, transform=None):
        self.data = images
        self.activations = activations
        self.transform = transform
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        image = self.data[idx]
        activation_score = self.activations[idx]
        
        if isinstance(image, torch.Tensor):
            image = image.permute(1, 2, 0).numpy()
        elif image.ndim == 3 and image.shape[0] in {1, 3}:
            image = image.transpose(1, 2, 0)  
        
        if self.transform:
            image = self.transform(image)
        return image, activation_score

    
    
transform = transforms.Compose([
    #transforms.Resize((32, 32)),
    #transforms.Resize((64, 64)),
    transforms.ToTensor(),
    #transforms.Resize((128, 128))
    #transforms.Normalize((0.0,), (2,))
])

testset = CustomDataset(images, test_feature_maps_tensor, transform=transform)

testloader = DataLoader(testset, batch_size=100, shuffle=False, num_workers=2)
```

### Diversity Generation

```python
device = torch.device("cuda:2" if torch.cuda.is_available() else "cpu")
num_top_images = 6

activations = []
indices = []

with torch.no_grad():
    for idx, (data, activation) in enumerate(testloader):
        activation = activation
        unit_activations = activation[:, unit_idx]
        activations.append(unit_activations.cpu())
        batch_indices = torch.arange(idx * testloader.batch_size, idx * testloader.batch_size + len(data))
        indices.append(batch_indices)

# Concatenate activations and indices
activations = torch.cat(activations)
indices = torch.cat(indices)

# Sort activations to get top images
sorted_indices = torch.argsort(activations, descending=False)
top_indices = indices[sorted_indices[:num_top_images]]

top_latents = all_latents[top_indices]
print(top_indices)
```

```python
print(top_latents)
```

```python
# Set Radius
latent_dim = 1024
num_units = 1  # Assuming you have 32 units per group
num_groups = latent_dim // num_units  # Number of groups (6 in your case)

# Create the prior latent vector with specific dimensions set to 1
prior_latent = -1 * torch.ones(num_groups)
    
constrained_latents_idx = range(num_groups)
# Calculate distances between top latents and the prior latent
chosen_top_latents = top_latents[:, constrained_latents_idx]

distances = []
for latent in chosen_top_latents:
    distance = torch.norm(latent - prior_latent)
    distances.append(distance.item())

# Compute the average radius
radius = np.mean(distances)
print(f"Calculated Radius: {radius}")

```

```python

def generate_orthogonal_directions(n, d, device):
    """
    Generates n orthogonal unit vectors in d-dimensional space.
    """
    directions = []
    for _ in range(n):
        # Generate a random vector
        vec = torch.randn(d).to(device)
        # Orthogonalize against existing directions
        for existing in directions:
            proj = torch.dot(vec, existing) * existing
            vec = vec - proj
        # Normalize the vector
        vec = vec / (vec.norm() + 1e-10)
        directions.append(vec)
    return torch.stack(directions)  # Shape: [n, d]
```

```python
# Interpolation to a random pic's latent
num_new_latents = num_top_images

# Get a random existing latent to fill the remaining dimensions
random_idx = np.random.choice(len(all_latents))
random_latent = all_latents[random_idx]



# Dimensions related to unit_idx
unit_dims = range(num_groups)

directions = 1.5 * radius * generate_orthogonal_directions(num_new_latents, len(unit_dims), device)
print(directions.shape)
new_latents = []
for i in range(num_new_latents):
    new_latent = torch.zeros(latent_dim)

    
    for idx, dim in enumerate(unit_dims):
        new_latent[dim] = prior_latent[idx] + directions[i][idx]
    # Fill the remaining dimensions with the random latent

    for i in range(latent_dim):
        if i not in unit_dims:
            new_latent[i] = random_latent[i]
    
    new_latents.append(new_latent)


new_latents = torch.stack(new_latents)
print(new_latents[:, unit_dims].shape)
print(random_idx)
```

```python
# Generation
with torch.no_grad():
    vae.eval()
    model.eval()
    #new_latents[:,:] = 1
    
    new_recons = vae.decode(new_latents.to(device))
    new_recons = 2 * new_recons - 1  # Scale to [-1, 1] if needed

    # Prepare noisy input for DDPM
    batch_size = new_recons.size(0)
    x_t = model.online_network.compute_noisy_input(
        torch.randn_like(new_recons),  # Start from noise
        torch.randn_like(new_recons),
        torch.tensor([model.online_network.T - 1] * batch_size, device=device),
    )

    if isinstance(model.online_network, DDPMv2):
        x_t += new_recons

    # Generate high-quality images using DDPM
    new_generated_images = model(
        x_t,
        cond=new_recons,
        z=None,
        n_steps=evaluation_n_steps,
        checkpoints=[],
        ddpm_latents=None,
    )[str(evaluation_n_steps)]

    # Scale back to [0, 1] if needed
    new_recons = new_recons * 0.5 + 0.5
    new_generated_images = new_generated_images * 0.5 + 0.5
```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import torchvision.models as models
import torchvision.transforms as transforms

# Load the pretrained Inception V3 model with aux_logits=True
inception_v3 = models.inception_v3(pretrained=True, aux_logits=True)
inception_v3.fc = nn.Identity()  # Remove the classification head
# Optionally, remove the auxiliary classifier
inception_v3.AuxLogits = nn.Identity()
inception_v3 = inception_v3.to(device)
inception_v3.eval()

# Define the preprocessing transformations
preprocess = transforms.Compose([
    transforms.Resize(299),
    transforms.CenterCrop(299),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],  # ImageNet mean
        std=[0.229, 0.224, 0.225]    # ImageNet std
    )
])

# Assuming top_images and new_generated_images are already loaded as tensors
# and have shape (N, C, H, W). If not, load them accordingly.

# Example:
# top_images = ...  # Tensor of shape (N, 3, 299, 299)
# new_generated_images = ...  # Tensor of shape (M, 3, 299, 299)

# Apply preprocessing if necessary
# If your images are PIL images or not yet normalized, apply the preprocess
# If they are already tensors and properly normalized, you can skip this step

# top_images = preprocess(top_images)
# new_generated_images = preprocess(new_generated_images)

# Create TensorDatasets
top_dataset = TensorDataset(top_images)
gen_dataset = TensorDataset(new_generated_images)

# Create DataLoaders
batch_size = 32  # Adjust based on your GPU memory
top_loader = DataLoader(top_dataset, batch_size=batch_size, shuffle=False)
gen_loader = DataLoader(gen_dataset, batch_size=batch_size, shuffle=False)

# Function to extract features, handling auxiliary outputs
def extract_features(model, loader, device):
    all_features = []
    with torch.no_grad():
        for batch in loader:
            images = batch[0].to(device)
            outputs = model(images)
            if isinstance(outputs, tuple):
                features = outputs[0]  # Main output
            else:
                features = outputs
            all_features.append(features.cpu())
    return torch.cat(all_features, dim=0)

# Extract and compute variance for original images
top_features = extract_features(inception_v3, top_loader, device)
ori_variances = top_features.var(dim=0)
print(f"Original variance: {ori_variances.mean().item()}")

# Extract and compute variance for generated images
gen_features = extract_features(inception_v3, gen_loader, device)
gen_variances = gen_features.var(dim=0)
print(f"New Variance: {gen_variances.mean().item()}")
```

### Optimize with Latent

```python
from torch.nn import Parameter
import torch.optim as optim
import math

def to_np(tensor_input):
    return tensor_input.detach().cpu().numpy()


def KLD_loss(mu, logvar, device, prior_mu):
    prior_var = (torch.ones(1, mu.size()[1]) * 0.01).to(device)  # prior_std=0.1, prior_var=0.01
    prior_logvar = torch.log(prior_var)
    prior_mu = (torch.ones_like(mu) * prior_mu).to(device)

    # Compute the KL divergence
    KLD = 0.5 * torch.mean(
        prior_logvar - logvar
        + (torch.exp(logvar) + (mu - prior_mu) ** 2) / prior_var
        - 1
    )
    return KLD

def optimize_activation_with_latent(net,
                                        vae,
                                        layer_name, 
                                        unit_idx,
                                        per_feature_num_latent = 6,
                                        num_features = 16,
                                        num_iterations=5000, 
                                        KLD_weight = 1.0,
                                        target = "max",
                                        mode = "reg",
                                        learning_rate=0.1, 
                                        l2_reg=1e-3, 
                                        log_reg = 1e-3,
                                        tv_reg=1e-3, 
                                        noise_coef = 0.1, 
                                        energy_scale = 1.0, 
                                        decay_steps = 100, 
                                        decay_factor = 0.9,
                                        init = 'normal', # ['uniform', 'data']
                                        init_data = None,
                                        use_regularization = False,
                                        use_projection = False,
                                        use_likelihood = False,
                                        use_energy = False):
    net.eval()
    vae.eval()
    
    device = next(net.parameters()).device
    #print("Energy Scale is: ", energy_scale)
    if init == 'normal':
        input_latent = torch.randn((1, 100), requires_grad=True, device=device) # Change to single channel
    elif init == 'uniform':
        input_latent = (torch.rand((1, 100), requires_grad=True, device=device) - 0.5) / 0.5 # Change to single channel
    elif init == 'data':
        init_mu, logvar = vae.encode(init_data)

        """
        if target == "max":
            for ith_latent in range(per_feature_num_latent):
                pre_mu[:,ith_latent*num_features+unit_idx[0]] = 1
        else:
            for ith_latent in range(per_feature_num_latent):
                pre_mu[:,ith_latent*num_features+unit_idx[0]] = -1      
        """
        input_latent = Parameter(torch.empty(init_mu.size(), device=device), requires_grad=True)
        input_latent.data.copy_(init_mu)
        mu_init = Parameter(torch.empty(init_mu.size(), device=device), requires_grad=False)
        mu_init.data.copy_(init_mu)

    else:
        raise ValueError("init type is not supported")
    
    optimizer = optim.Adam([input_latent], lr=learning_rate)

    optimizer.zero_grad()
    
    target_activation = [None]
    
    def hook(module, input, output):
        nonlocal target_activation
        if isinstance(unit_idx, tuple):
            #target_activation[0] = output[:, unit_idx[0], unit_idx[1], unit_idx[2]]
            target_activation[0] = output[:, unit_idx[0]].mean(-1).mean(-1)
        elif isinstance(unit_idx, list):
            unit_idx_item = unit_idx[0]
            target_activation[0] = output[0, unit_idx_item[0], unit_idx_item[1], unit_idx_item[2]]
            for unit_idx_item in unit_idx[1:]:
                target_activation[0] += output[0, unit_idx_item[0], unit_idx_item[1], unit_idx_item[2]]
        else:
            target_activation[0] = output[0, unit_idx]

    layers = {
        'layer1_0_conv1': net.encoder.model.convnet.layer1[0].conv1,
        'layer1_0_conv2': net.encoder.model.convnet.layer1[0].conv2,
        'layer2_2_conv2': net.encoder.model.convnet.layer2[2].conv2,
    }
    layer = layers[layer_name]
    hook_handle = layer.register_forward_hook(hook)
    
    mask = torch.zeros_like(init_mu)
    for ith_latent in range(per_feature_num_latent):
        mask[:,ith_latent*num_features+unit_idx[0]] = 1

    
    step_size = learning_rate
    for i in range(num_iterations):
        
        if i % decay_steps == 0 and i > 0:
            step_size *= decay_factor
        
        # pre_mu_new_params = []
        # old_index = 0
        # for ith_latent in range(per_feature_num_latent):
        #     pre_mu_new_params.append(pre_mu[:,old_index:ith_latent*num_features+unit_idx[0]])
        #     # print(pre_mu[:,old_index:ith_latent*num_features+unit_idx[0]].shape)
        #     pre_mu_new_params.append(input_latent[:,ith_latent].unsqueeze(-1))
        #     # print(input_latent[:,ith_latent].unsqueeze(-1).shape)
        #     old_index = ith_latent*num_features+unit_idx[0]+1
        # pre_mu_new_params.append(pre_mu[:,old_index:])
        # print(torch.cat(pre_mu_new_params, dim = 1).shape)
        # post_mu = vae.fc_bn_mu(torch.cat(pre_mu_new_params, dim = -1))
        # import pdb; pdb.set_trace()
        
        # post_mu = vae.fc_bn_mu(input_latent)
        
        temp_mu = mu_init * (1 - mask) + input_latent * mask
        
        input_image = vae.decode(temp_mu)

        _ = net.encoder(input_image)

        # When constrained:
        if mode == "reg":
            if target == "max":
                KLD = KLD_loss(mu = input_latent[:,[i * num_features + unit_idx[0] for i in range(per_feature_num_latent)]],
                       logvar = logvar[:,[i * num_features + unit_idx[0] for i in range(per_feature_num_latent)]].detach(),
                       device = device,
                       prior_mu = 1)
            else:
                KLD = KLD_loss(mu = input_latent[:,[i * num_features + unit_idx[0] for i in range(per_feature_num_latent)]],
                       logvar = logvar[:,[i * num_features + unit_idx[0] for i in range(per_feature_num_latent)]].detach(),
                       device = device,
                       prior_mu = -1)
        
        else:
            logvar = logvar.detach()
            mu = input_latent
            KLD = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())

        if target == "max":
            loss = -target_activation[0].mean() + KLD_weight * KLD
        else:
            loss = target_activation[0].mean() + KLD_weight * KLD
        if i == 0:
            pass
            #print("Init Activation score: ", target_activation[0].mean())
        
        if (i + 1) % 200 == 0:
            pass
            #print("Activation score: ", target_activation[0].mean())
            #print("KLD loss: ", KLD_weight * KLD)
        
        if use_regularization:
            l2_loss = l2_reg * torch.sum(input_image**2)
            loss += l2_loss
            
            tv_loss = tv_reg * (torch.sum(torch.abs(input_image[:, :, :-1, :] - input_image[:, :, 1:, :])) +
                            torch.sum(torch.abs(input_image[:, :, :, :-1] - input_image[:, :, :, 1:])))
            loss += tv_loss
        
        if use_likelihood:
            # Compute Gaussian log likelihood
            log_likelihood = -0.5 * math.log(2 * math.pi * image_std**2) - (input_image - image_mu).pow(2).mean() / (2 * image_std**2)
            loss += log_reg * (-log_likelihood)

        loss.backward()
        
        if use_energy:
            with torch.no_grad():
                # Update latent by descending along the gradient and adding Gaussian noise
                input_latent -= energy_scale * step_size * input_latent.grad
                #print(input_latent.grad)
                input_latent += step_size * 2 * torch.randn_like(input_latent)
            # optimizer.step()

            # Zero the gradients before the next step
            input_latent.grad.zero_()
        else:
            optimizer.step()
            optimizer.zero_grad()

        if use_projection:
            # projection step
            input_latent.data.clamp_(min=-1, max=1)
    
    scores_all = []
    
    scores_all.append(to_np(target_activation[0].mean()))
    input_image = input_image.detach()
    
    print('activation_score_LD_optimize:', to_np(target_activation[0].mean()))
    
    ######################################
    
    _ = net.encoder(init_data)
    
    scores_all.append(to_np(target_activation[0].mean()))
    
    print('activation_score_init:', to_np(target_activation[0].mean()))
    
    ######increase prior
    
    init_mu_increase = init_mu.clone()
    for ith_latent in range(per_feature_num_latent):
        init_mu_increase[:,ith_latent*num_features+unit_idx[0]] = 1
    
    input_image_increase = vae.decode(init_mu_increase)
    
    _ = net.encoder(input_image_increase)
    
    scores_all.append(to_np(target_activation[0].mean()))
    
    print('activation_score_increase_prior:', to_np(target_activation[0].mean()))
    
    input_image_increase = input_image_increase.detach()
    input_image_increase = (input_image_increase - input_image_increase.min()) / (input_image_increase.max() - input_image_increase.min())
    
    ######decrease prior
    
    init_mu_decrease = init_mu.clone()
    for ith_latent in range(per_feature_num_latent):
        init_mu_decrease[:,ith_latent*num_features+unit_idx[0]] = -1
    
    input_image_decrease = vae.decode(init_mu_decrease)
    
    _ = net.encoder(input_image_decrease)
    
    scores_all.append(to_np(target_activation[0].mean()))
    
    print('activation_score_decrease_prior:', to_np(target_activation[0].mean()))
    
    input_image_decrease = input_image_decrease.detach()
    input_image_decrease = (input_image_decrease - input_image_decrease.min()) / (input_image_decrease.max() - input_image_decrease.min())
        
    hook_handle.remove()

    return input_image, input_image_increase, input_image_decrease, scores_all
```
